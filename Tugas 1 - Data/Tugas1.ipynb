### 📘 UK 1 · Langkah 3 — Mencatat penerimaan berkas
**Apa yang dilakukan:** mencatat nama, ukuran, dan **SHA-256** file.
**Penjelasan:** SHA-256 adalah sidik jari file. Satu huruf berubah, sidik jarinya ikut berubah, jadi nanti bisa dibuktikan bahwa file sumber tidak diubah.

```python
sumber = Path(FILE_NAME)
assert sumber.exists(), f"File {FILE_NAME} tidak ditemukan. Ulangi Langkah 2."

ukuran_byte = sumber.stat().st_size
sha256_awal = hashlib.sha256(sumber.read_bytes()).hexdigest()

print("Nama berkas :", sumber.name)
print("Ukuran      :", ukuran_byte, "byte")
print("SHA-256     :", sha256_awal)
```

# BAGIAN UK 2 — Memvalidasi Data
Sekarang isi data diperiksa. Setiap baris diputuskan nasibnya: **bersih**, **dikarantina** (dikembalikan ke pemilik data), atau **disisihkan** (salinan persis yang tidak perlu dipakai dua kali). Tidak ada baris yang dihapus tanpa catatan.

### 📘 UK 2 · Langkah 1 — Membuka luaran UK 1
**Apa yang dilakukan:** membuka `01_data_hasil_pemasukan.csv`. Semua kolom dibaca sebagai **teks** agar nilai aslinya (mis. `abc`) tetap utuh sebagai arsip (`df_raw2`).
**Hasil yang benar:** `Data diterima dari UK 1: (20, 5)`

```python
INPUT_FILE = "01_data_hasil_pemasukan.csv"
df_raw2 = pd.read_csv(INPUT_FILE, dtype="string")
data = df_raw2.copy()          # tabel kerja UK 2

print("Data diterima dari UK 1:", data.shape)
```

### 📘 UK 2 · Langkah 2 — Menetapkan aturan validasi
**Apa yang dilakukan:** menulis aturan benar/salah per kolom di satu tempat.
| Kolom | Aturan |
|---|---|
| `customer_id` | wajib diisi; tidak boleh kembar dengan isi berbeda |
| `age` | angka 18 sampai 100 |
| `contract` | `Monthly` atau `Annual` |
| `monthly_charge` | angka, tidak negatif |
| `churn` | `Yes` atau `No` |

```python
required = ["customer_id", "age", "contract", "monthly_charge", "churn"]
UMUR_MIN, UMUR_MAX = 18, 100
KONTRAK_BOLEH = ["Monthly", "Annual"]
CHURN_BOLEH = ["Yes", "No"]
print("Aturan siap. Kolom wajib:", required)
```

### 📘 UK 2 · Langkah 3 — Merapikan format (normalisasi)
**Apa yang dilakukan:** menyeragamkan penulisan: spasi dibuang dan huruf awal dikapitalkan (`monthly` → `Monthly`). Angka yang tidak terbaca (`abc`) menjadi kosong agar ketahuan.
**Penjelasan:** ini hanya memperbaiki **salah tulis yang jelas**. Nilai yang salah *isinya* tidak ditebak. Perubahan format dicatat (`perbaikan_format`).
**Hasil yang benar:** perbaikan format pada **C0004** (`monthly` → `Monthly`) dan **C0018** (`annual` → `Annual`).

```python
data["contract"] = data["contract"].astype("string").str.strip().str.title()
data["churn"] = data["churn"].astype("string").str.strip().str.title()
data["age"] = pd.to_numeric(data["age"], errors="coerce").astype("float64")
data["monthly_charge"] = pd.to_numeric(data["monthly_charge"], errors="coerce").astype("float64")

perbaikan_format = []
for kol in ["contract", "churn"]:
    beda = df_raw2[kol].notna() & (df_raw2[kol] != data[kol]).fillna(False)
    for i in data.index[beda]:
        perbaikan_format.append({"index": i, "kolom": kol, "asli": df_raw2.at[i, kol], "baru": data.at[i, kol]})
perbaikan_format = pd.DataFrame(perbaikan_format)

print("Perbaikan format:", len(perbaikan_format))
perbaikan_format
```

### 📘 UK 2 · Langkah 4 — Enam pemeriksaan (hanya menandai)
**Apa yang dilakukan:** menandai `True` pada baris yang melanggar aturan. Belum ada baris yang dihapus.
**Penjelasan:** `isna()` mencari sel kosong; `duplicated(keep=False)` menandai **kedua** baris kembar; `between(18, 100)` memeriksa rentang; `isin([...])` memeriksa daftar; tanda `~` = "bukan".

```python
data["issue_missing_required"] = data[required].isna().any(axis=1)
data["issue_duplicate_id"] = data["customer_id"].notna() & data["customer_id"].duplicated(keep=False)
data["issue_invalid_age"] = data["age"].notna() & ~data["age"].between(UMUR_MIN, UMUR_MAX)
data["issue_invalid_contract"] = data["contract"].notna() & ~data["contract"].isin(KONTRAK_BOLEH)
data["issue_negative_charge"] = data["monthly_charge"].notna() & data["monthly_charge"].lt(0)
data["issue_invalid_churn"] = data["churn"].notna() & ~data["churn"].isin(CHURN_BOLEH)

issue_cols = ["issue_missing_required", "issue_duplicate_id", "issue_invalid_age",
              "issue_invalid_contract", "issue_negative_charge", "issue_invalid_churn"]

acuan = {"issue_missing_required": 4, "issue_duplicate_id": 2, "issue_invalid_age": 2,
         "issue_invalid_contract": 1, "issue_negative_charge": 1, "issue_invalid_churn": 1}
hasil_cek = pd.DataFrame({
    "pemeriksaan": issue_cols,
    "jumlah_gagal": [int(data[c].sum()) for c in issue_cols],
    "angka_acuan": [acuan[c] for c in issue_cols],
})
hasil_cek["cocok?"] = hasil_cek["jumlah_gagal"].eq(hasil_cek["angka_acuan"]).map({True: "COCOK", False: "BEDA"})
hasil_cek
```

### 📘 UK 2 · Langkah 5 — Memutuskan nasib tiap baris
**Apa yang dilakukan:** membedakan dua jenis baris kembar.
**Penjelasan:**
* **Salinan persis** (semua isi sama, mis. C0015 baris 15 dan 16): baris pertama dipakai, baris berikutnya **disisihkan** karena hanya pengulangan.
* **ID sama tetapi isi berbeda:** tidak bisa ditebak mana yang benar, jadi **dikarantina**.
* Baris dengan masalah lain (kosong, umur salah, kontrak salah, dst.) **dikarantina**.
* Baris lolos semua = **bersih**.
**Hasil yang benar:** Bersih **10**, Dikarantina **9**, Disisihkan **1**.

```python
# Salinan persis: semua kolom sama. Baris pertama dipertahankan, salinan berikutnya disisihkan.
salinan_persis = data.duplicated(subset=required, keep=False) & data["customer_id"].notna()
dup_berikutnya = salinan_persis & data.duplicated(subset=required, keep="first")
# ID kembar tetapi isi berbeda -> tidak bisa ditebak -> karantina
data["issue_duplicate_conflict"] = data["issue_duplicate_id"] & ~salinan_persis

cols_karantina = ["issue_missing_required", "issue_duplicate_conflict", "issue_invalid_age",
                  "issue_invalid_contract", "issue_negative_charge", "issue_invalid_churn"]
ada_masalah = data[cols_karantina].any(axis=1)

data["keputusan"] = "Bersih"
data.loc[ada_masalah, "keputusan"] = "Dikarantina"
data.loc[dup_berikutnya & ~ada_masalah, "keputusan"] = "Disisihkan"

n_bersih = int((data["keputusan"] == "Bersih").sum())
n_karantina = int((data["keputusan"] == "Dikarantina").sum())
n_sisih = int((data["keputusan"] == "Disisihkan").sum())

print("Baris diterima:", len(data))
print("Bersih        :", n_bersih, "(acuan 10)")
print("Dikarantina   :", n_karantina, "(acuan 9)")
print("Disisihkan    :", n_sisih, "(acuan 1)")
print("Rekonsiliasi  :", n_bersih, "+", n_karantina, "+", n_sisih, "=", n_bersih + n_karantina + n_sisih,
      "->", "COCOK" if n_bersih + n_karantina + n_sisih == len(data) else "TIDAK COCOK")
```

### 📘 UK 2 · Langkah 6 — Mencatat semua keputusan (catatan validasi)
**Apa yang dilakukan:** menulis satu catatan untuk setiap keputusan: dikarantina, disisihkan, dan diperbaiki, dengan alasan dari **nilai asli**.
**Hasil yang benar:** **12 temuan** = 9 dikarantina + 1 disisihkan + 2 diperbaiki.

```python
def pesan_missing(i):
    bagian = []
    for kol in required:
        if pd.isna(data.at[i, kol]):
            asli = df_raw2.at[i, kol]
            bagian.append(f"{kol} kosong" if pd.isna(asli) else f"{kol} bukan angka (nilai asli: '{asli}')")
    return "; ".join(bagian)

PESAN = {
    "issue_missing_required": ("Nilai wajib kosong/tidak terbaca", pesan_missing),
    "issue_duplicate_conflict": ("ID kembar, isi berbeda", lambda i: f"customer_id {data.at[i, 'customer_id']} muncul lebih dari sekali dengan isi berbeda"),
    "issue_invalid_age": ("Umur di luar rentang", lambda i: f"age {df_raw2.at[i, 'age']} di luar rentang {UMUR_MIN}-{UMUR_MAX}"),
    "issue_invalid_contract": ("Kontrak tidak baku", lambda i: f"contract '{df_raw2.at[i, 'contract']}' bukan Monthly/Annual"),
    "issue_negative_charge": ("Tagihan negatif", lambda i: f"monthly_charge negatif ({df_raw2.at[i, 'monthly_charge']})"),
    "issue_invalid_churn": ("Churn tidak baku", lambda i: f"churn '{df_raw2.at[i, 'churn']}' bukan Yes/No"),
}

def id_tampil(i):
    return "(kosong)" if pd.isna(df_raw2.at[i, "customer_id"]) else df_raw2.at[i, "customer_id"]

catatan = []
alasan = {}
for i in data.index:
    keputusan = data.at[i, "keputusan"]
    if keputusan == "Dikarantina":
        teks_baris = []
        for c in cols_karantina:
            if data.at[i, c]:
                judul, fungsi = PESAN[c]
                teks = fungsi(i)
                teks_baris.append(teks)
                catatan.append({"no_baris_data": i + 1, "customer_id": id_tampil(i), "pemeriksaan": judul,
                                "detail_temuan": teks, "keputusan": "Dikarantina"})
        alasan[i] = "; ".join(teks_baris)
    elif keputusan == "Disisihkan":
        pertama = data.index[(data["customer_id"] == data.at[i, "customer_id"]) & ~dup_berikutnya][0] + 1
        catatan.append({"no_baris_data": i + 1, "customer_id": id_tampil(i), "pemeriksaan": "Duplikat persis",
                        "detail_temuan": f"salinan persis baris {pertama}; baris {pertama} dipakai", "keputusan": "Disisihkan"})
    else:
        for _, r in perbaikan_format[perbaikan_format["index"] == i].iterrows():
            catatan.append({"no_baris_data": i + 1, "customer_id": id_tampil(i), "pemeriksaan": "Format tidak seragam",
                            "detail_temuan": f"{r['kolom']} '{r['asli']}' diseragamkan menjadi '{r['baru']}'", "keputusan": "Diperbaiki"})

catatan = pd.DataFrame(catatan).sort_values("no_baris_data").reset_index(drop=True)
print("Jumlah temuan:", len(catatan), "(acuan 12)")
print(catatan["keputusan"].value_counts().to_string())
catatan
```

### 📘 UK 2 · Langkah 7 — Menyusun tabel luaran
**Apa yang dilakukan:** menyiapkan tiga tabel.
* **Bersih:** 5 kolom yang sudah rapi, siap dipakai tahap berikutnya.
* **Karantina:** memakai **nilai asli** plus alasan, agar pemilik data tahu persis apa yang harus dikoreksi.
* **Catatan validasi:** semua keputusan dari Langkah 6.

```python
bersih_out = data.loc[data["keputusan"] == "Bersih", required].copy()
bersih_out["age"] = bersih_out["age"].astype(int)
bersih_out["monthly_charge"] = bersih_out["monthly_charge"].astype(int)

def nilai_asli(v):
    if pd.isna(v):
        return None
    try:
        angka = float(v)
        return int(angka) if angka.is_integer() else angka
    except ValueError:
        return str(v)

idx_k = data.index[data["keputusan"] == "Dikarantina"]
karantina_out = pd.DataFrame({
    "no_baris_data": [i + 1 for i in idx_k],
    "customer_id": [nilai_asli(df_raw2.at[i, "customer_id"]) for i in idx_k],
    "age": [nilai_asli(df_raw2.at[i, "age"]) for i in idx_k],
    "contract": [nilai_asli(df_raw2.at[i, "contract"]) for i in idx_k],
    "monthly_charge": [nilai_asli(df_raw2.at[i, "monthly_charge"]) for i in idx_k],
    "churn": [nilai_asli(df_raw2.at[i, "churn"]) for i in idx_k],
    "alasan_karantina": [alasan[i] for i in idx_k],
}, dtype="object")   # object agar angka tetap utuh (17, bukan 17.0)
print("Tabel luaran siap:", bersih_out.shape, karantina_out.shape, catatan.shape)
```

### 📘 UK 2 · Langkah 8 — Menyimpan tiga berkas output (CSV)
**Apa yang dilakukan:** menyimpan hasil UK 2 sebagai tiga file CSV. Inilah yang dikumpulkan di folder Tugas 1.
**Hasil yang benar:** tiga nama file tercetak.

```python
bersih_out.to_csv("data_pelanggan_bersih.csv", index=False)
karantina_out.to_csv("data_karantina.csv", index=False)
catatan.to_csv("catatan_validasi.csv", index=False)

print("File hasil UK 2 tersimpan:")
for nama in ["data_pelanggan_bersih.csv", "data_karantina.csv", "catatan_validasi.csv"]:
    print("-", nama)
```

### 📘 UK 2 · Langkah 9 — Cek ulang file yang tersimpan
**Apa yang dilakukan:** membuka lagi file dari penyimpanan dan menjalankan aturan yang sama pada data bersih. Ini bukti isinya benar, bukan sekadar percaya pada variabel di memori.
**Hasil yang benar:** `Bersih: 10 baris`, `Karantina: 9 baris`, `Catatan: 12 temuan`, `cek ulang: 0 cacat`.
📸 *Bahan screenshot UK 2.*

```python
cek_bersih = pd.read_csv("data_pelanggan_bersih.csv")
cek_karantina = pd.read_csv("data_karantina.csv")
cek_catatan = pd.read_csv("catatan_validasi.csv")

cacat = (cek_bersih[required].isna().any(axis=1)
         | ~cek_bersih["age"].between(UMUR_MIN, UMUR_MAX)
         | ~cek_bersih["contract"].isin(KONTRAK_BOLEH)
         | (cek_bersih["monthly_charge"] < 0)
         | ~cek_bersih["churn"].isin(CHURN_BOLEH)
         | cek_bersih["customer_id"].duplicated(keep=False)).sum()

print("Bersih    :", len(cek_bersih), "baris (acuan 10) ->", "COCOK" if len(cek_bersih) == 10 else "BEDA")
print("Karantina :", len(cek_karantina), "baris (acuan 9)  ->", "COCOK" if len(cek_karantina) == 9 else "BEDA")
print("Catatan   :", len(cek_catatan), "temuan (acuan 12) ->", "COCOK" if len(cek_catatan) == 12 else "BEDA")
print("Cek ulang data bersih:", int(cacat), "cacat", "->", "COCOK" if cacat == 0 else "BEDA")
```
